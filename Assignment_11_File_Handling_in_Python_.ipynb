2. Write to a Text File  

```python
text = input("Enter text to write: ")

# Write Mode
with open("user_data.txt", "w") as file:
    file.write(text + "\n")

print("Data written successfully.")

more_text = input("Enter text to append: ")

# Append Mode
with open("user_data.txt", "a") as file:
    file.write(more_text + "\n")

print("\nContents of user_data.txt:\n")

with open("user_data.txt", "r") as file:
    print(file.read())
```

3. Read File Using File Object (Stream)

Using read()

```python
with open("Jala Academy.txt", "r") as file:
    print("Using read():\n")
    print(file.read())
```

Using readline()

```python
with open("Jala Academy.txt", "r") as file:
    print("Using readline():\n")
    print(file.readline())
    print(file.readline())
```

Using readlines()

```python
with open("Jala Academy.txt", "r") as file:
    print("Using readlines():\n")

    lines = file.readlines()

    for line in lines:
        print(line, end="")
```

Comparison

```python
print("Comparison:")
print("read()      -> Reads the entire file.")
print("readline()  -> Reads one line at a time.")
print("readlines() -> Reads all lines and returns a list.")
```

4. Random Access File Reading  

```python
with open("Jala Academy.txt", "r") as file:

    file.seek(100)

    print("Cureading from Position 100:\n")

    print(file.read())
```

5. Read from a Specific Index

```python
index = int(input("Enter starting index: "))
count = int(input("Enter number of characters: "))

with open("Jala Academy.txt", "r") as file:
    file.seek(index)
    data = file.read(count)

print("\nCharacters Read:\n")
print(data)
```

6. Check File Permissions

```python
import os

filename = "Jala Academy.txt"

if os.access(filename, os.R_OK):
    print("Read Permission: Yes")
else:
    print("Read Permission: No")

if os.access(filename, os.W_OK):
    print("Write Permission: Yes")
else:
    print("Write Permission: No")
```

7. Count Words, Lines, and Characters

```python
with open("Jala Academy.txt", "r") as file:
    content = file.read()

characters = len(content)
words = len(content.split())
lines = len(content.splitlines())

print("Number of Lines      :", lines)
print("Number of Words      :", words)
print("Number of Characters :", characters)
```

8. Copy File Content

```python
with open("Jala Academy.txt", "r") as source:
    content = source.read()

with open("copy_Jala Academy.txt", "w") as destination:
    destination.write(content)

print("File copied successfully.")

print("\nContents of copy_Jala Academy.txt:\n")

with open("copy_Jala Academy.txt", "r") as file:
    print(file.read())
```

9. Append Data with Timestamp

```python
from datetime import datetime

text = input("Enter text to append: ")

current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

with open("sample.txt", "a") as file:
    file.write(f"\n[{current_time}] {text}")

print("Data appended successfully.\n")

print("Updated File Content:\n")

with open("Jala Academy.txt", "r") as file:
    print(file.read())
```
