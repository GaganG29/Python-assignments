#Assignment-14: Exception Handling in Python

1. Generate an Exception

```python
a = 10
b = 0
result = a / b
print(result)
```

2. Handle the Exception

```python
a = 10
b = 0

try:
    result = a / b
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

3. Multiple Except Blocks

```python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    result = num1 / num2
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
except ValueError:
    print("Error: Please enter valid integers.")
```

4. Raise an Exception Manually

```python
def check_age(age):
    if age < 18:
        raise Exception("Age must be 18 or above.")
    print("Eligible.")
try:
    check_age(16)
except Exception as e:
    print(e)
```

5. Function That Raises Exception

Without Handling

```python
def show_error():
    raise Exception("This is a custom exception.")
show_error()
```

With Handling

```python
def show_error():
    raise Exception("This is a custom exception.")
try:
    show_error()
except Exception as e:
    print("Handled:", e)
```

6. Create Your Own Exception

```python
class InsufficientBalanceError(Exception):
    pass

balance = 500
withdraw = 700

try:
    if withdraw > balance:
        raise InsufficientBalanceError("Insufficient Balance.")
    balance -= withdraw
    print("Remaining Balance:", balance)

except InsufficientBalanceError as e:
    print(e)
```

7. Using finally

```python
file = None
try:
    file = open("sample.txt", "w")
    file.write("Hello Python")
    print("Data written successfully.")
except Exception as e:
    print(e)
finally:
    if file:
        file.close()
    print("File closed.")
```

8. File Not Found

```python
try:
    file = open("unknown.txt", "r")

except FileNotFoundError:
    print("File not found.")
```

9. Type Error

```python
try:
    result = 10 + "Python"
    print(result)

except TypeError:
    print("Type Error occurred.")
```

10. Attribute Error

```python
class Student:
    def __init__(self):
        self.name = "Riya"
obj = Student()
try:
    print(obj.age)
except AttributeError:
    print("Attribute does not exist.")
```

11. Index Error

```python
numbers = [10, 20, 30]
try:
    print(numbers[5])
except IndexError:
    print("Index out of range.")
```

12. Use else Block

```python
try:
    num1 = 20
    num2 = 5
    result = num1 / num2
except ZeroDivisionError:
    print("Division by zero.")
else:
    print("Result:", result)
```

13. Logging Errors

```python
import logging
logging.basicConfig(filename="error.log", level=logging.ERROR)
try:
    result = 10 / 0
except Exception as e:
    logging.error("An error occurred: %s", e)
    print("Error logged successfully.")
```

14. Input Validation System

```python
while True:
    try:
        number = int(input("Enter an integer: "))
        print("You entered:", number)
        break
    except ValueError:
        print("Invalid input. Please enter a valid integer.")
```
