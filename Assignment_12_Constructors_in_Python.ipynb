#Assignment-12: Constructors in Python

1. Default and Parameterized Constructors

```python
# 1. Default and Parameterized Constructors

class Student:
    def __init__(self, name="Unknown", age=0):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age :", self.age)
        print("-" * 20)

# Default constructor
obj1 = Student()
# One-argument constructor
obj2 = Student("Sathvika")
# Two-argument constructor
obj3 = Student("Ashwini", 22)

obj1.display()
obj2.display()
obj3.display()
```

2. Calling Parent Constructor

```python
class College:
    def __init__(self, college_name):
        self.college_name = college_name
        print("College Constructor Called")

class Student(College):
    def __init__(self, college_name, student_name):
        super().__init__(college_name)
        self.student_name = student_name
        print("Student Constructor Called")

    def display(self):
        print("College Name:", self.college_name)
        print("Student Name:", self.student_name)

obj = Student("CMR Engineering College", "Sathvika")
obj.display()
```

3. Simulating Access Levels

```python
class Demo:
    def __init__(self):
        self.public = "Public Variable"
        self._protected = "Protected Variable"
        self.__private = "Private Variable"
obj = Demo()
print("Public:", obj.public)
print("Protected:", obj._protected)
try:
    print(obj.__private)
except AttributeError:
    print("Private variable cannot be accessed directly.")
print("Private using name mangling:", obj._Demo__private)
```

4. Constructor Attributes Concept

```python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course
obj = Student("Sathvika", 21, "B.Tech IT")
print("Name :", obj.name)
print("Age  :", obj.age)
print("Course:", obj.course)
```

5. Using __str__() for Better Output

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"Student(Name={self.name}, Age={self.age})"
obj = Student("Sathvika", 21)
print(obj)
```

6. Constructor with *args

```python
class Numbers:
    def __init__(self, *args):
        self.values = args
    def display(self):
        print("Values:", self.values)
obj = Numbers(10, 20, 30, 40, 50)
obj.display()
```

7. Real-World Example

```python
class Employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary
    def display(self):
        print("Employee Name :", self.name)
        print("Employee ID   :", self.emp_id)
        print("Salary        :", self.salary)
        print("-" * 30)
emp1 = Employee("Sathvika", 101, 35000)
emp2 = Employee("Ashwini", 102, 42000)
emp3 = Employee("Renu", 103, 50000)

emp1.display()
emp2.display()
emp3.display()
```
