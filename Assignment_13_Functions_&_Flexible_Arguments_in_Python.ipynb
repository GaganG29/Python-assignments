#Assignment-13: Functions & Flexible Arguments in Python

1. Function with Parameters

```python
def add_numbers(num1, num2):
    return num1 + num2
result = add_numbers(15, 25)
print("Sum =", result)
```

2. Default Parameters

```python
def greet(name, message="Hello"):
    print(message + ", " + name)

greet("Sathvika")
greet("Ashwini", "Good Morning")
```

*3*. Keyword Arguments Usage

```python
def person_details(name, age, city):
    print("Name:", name)
    print("Age :", age)
    print("City:", city)

person_details(age=22, city="Hyderabad", name="Priya")
print()

person_details(city="Bengaluru", name="Alie", age=25)
```

4. Using *args

```python
def find_sum(*numbers):
    return sum(numbers)
print("Sum =", find_sum(10, 20, 30, 40, 50))
```

5. Using **kwargs

```python
def display_details(**details):
    for key, value in details.items():
        print(key + ":", value)
display_details(Name="Bob", Age=23, Course="Python", City="Hyderabad")
```

6. Flexible Function

```python
def check_input(value):
    if isinstance(value, int) or isinstance(value, float):
        print("Square =", value * value)
    elif isinstance(value, str):
        print("Uppercase =", value.upper())
    else:
        print("Unsupported Data Type")
check_input(8)
check_input("python")
```

7. Advanced Function

```python
def calculate_bill(price=100, *items, discount=0, **extra_items):
    total = price
    for item in items:
        total += item
    for value in extra_items.values():
        total += value
    total = total - discount
    print("Final Bill =", total)
calculate_bill(500, 100, 200, 150, discount=50, Delivery=40, Packing=20)
```

8. Lambda Functions

```python
add = lambda a, b: a + b
square = lambda x: x * x
print("Addition =", add(12, 18))
print("Square =", square(9))
```

9. Map Function

```python
numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x * x, numbers))
print("Original List:", numbers)
print("Squared List:", squares)
```

*10*. Filter Function

```python
numbers = [10, 15, 20, 25, 30, 35, 40]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print("Original List:", numbers)
print("Even Numbers:", even_numbers)
```
